## Question 6: Crop the Central 200 × 200 Pixel Area

## Write a Python code snippet to crop the central 200 × 200 pixel area of an image and display it using Matplotlib.

```python
# Read the original flower image
img = cv2.imread(r"C:\Users\hpw\Downloads\IPP_ASSIGNMENT_2\flower.jpg")

# Check whether the image was loaded successfully
if img is not None:

    # Get the height and width of the image
    height, width = img.shape[:2]
    # Calculate the horizontal centre of the image
    x = width // 2
    # Calculate the vertical centre of the image
    y = height // 2
    # Crop a 200 by 200 pixel area around the centre
    cropped = img[y-100:y+100, x-100:x+100]
    # Convert the cropped image from BGR to RGB for correct display
    cropped = cv2.cvtColor(cropped, cv2.COLOR_BGR2RGB)
    # Display the cropped image
    plt.imshow(cropped)
    # Set the title of the displayed image
    plt.title("Cropped Image")
    # Hide the x-axis and y-axis
    plt.axis("off")
    # Show the cropped image
    plt.show()

# Display an error message if the image is not found
else:
    print("Image not found. Check the file path.")
```

## Question 8: Flip an Image Horizontally

## Write a Python statement to flip the image `car.jpg` left-to-right using OpenCV and display the result.

```python
# Read the car image from the assignment folder
img = cv2.imread(r"C:\Users\hpw\Downloads\IPP_ASSIGNMENT_2\car.jpg")

# Check whether the image was loaded successfully
if img is not None:
    # Flip the image horizontally from left to right
    flipped = cv2.flip(img, 1)
    # Convert the original image from BGR to RGB
    original = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
    # Convert the flipped image from BGR to RGB
    flipped = cv2.cvtColor(flipped, cv2.COLOR_BGR2RGB)
    # Set the size of the figure used to display both images
    plt.figure(figsize=(10, 4)) 
    # Create the first subplot in a layout of 1 row and 2 columns
    plt.subplot(1, 2, 1)
    # Display the original image in the first subplot
    plt.imshow(original)
    # Set the title for the original image
    plt.title("Original Image")
    # Hide the axes of the first subplot
    plt.axis("off")
    # Create the second subplot in the same figure
    plt.subplot(1, 2, 2)
    # Display the flipped image in the second subplot
    plt.imshow(flipped)# Set the title for the flipped image
    plt.title("Flipped Image")
    # Hide the axes of the second subplot
    plt.axis("off")
    # Display both images together
    plt.show()

# Display an error message if the image is not found
else:
    print("Image not found. Check the file path.")
```

## Question 9: Convert an Image to HSV

## Write a Python program that converts an image from RGB to HSV colour space, and display each channel (Hue, Saturation, Value) separately using Matplotlib.

```python
# Read the image from the assignment folder
img = cv2.imread(r"C:\Users\hpw\Downloads\IPP_ASSIGNMENT_2\img2.jpg")

# Check whether the image was loaded successfully
if img is not None:
    # Convert the image from BGR colour space to HSV colour space
    hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)
    # Separate the HSV image into Hue, Saturation, and Value channels
    h, s, v = cv2.split(hsv)
    # Set the size of the figure used to display the three channels
    plt.figure(figsize=(5, 3))
    # Create the first subplot for the Hue channel
    plt.subplot(1, 3, 1)
    # Display the Hue channel in grayscale
    plt.imshow(h, cmap="gray")
    # Set the title for the Hue channel
    plt.title("Hue")
    # Hide the axes of the first subplot
    plt.axis("off")
    # Create the second subplot for the Saturation channel
    plt.subplot(1, 3, 2)
    # Display the Saturation channel in grayscale
    plt.imshow(s, cmap="gray")
    # Set the title for the Saturation channel
    plt.title("Saturation")
    # Hide the axes of the second subplot
    plt.axis("off")
    # Create the third subplot for the Value channel
    plt.subplot(1, 3, 3)
    # Display the Value channel in grayscale
    plt.imshow(v, cmap="gray")
    # Set the title for the Value channel
    plt.title("Value")
    # Hide the axes of the third subplot
    plt.axis("off")
    # Display all three channels together
    plt.show()

# Display an error message if the image is not found
else:
    print("Image not found. Check the file path.")
```

## Question 10: Calculate Mean Squared Error (MSE)

## Write a short Python code snippet to compute the Mean Squared Error (MSE) between two grayscale images, `img1.jpg` and `img2.jpg`.

```python
# Read the first image in grayscale
img1 = cv2.imread(r"C:\Users\hpw\Downloads\IPP_ASSIGNMENT_2\flower.jpg",0)

# Read the second image in grayscale
img2 = cv2.imread(r"C:\Users\hpw\Downloads\IPP_ASSIGNMENT_2\img2.jpg",0)

# Check whether both images were loaded successfully
if img1 is not None and img2 is not None:
    # Resize the second image to match the first image
    img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

    # Convert images to float
    img1 = img1.astype(float)
    img2 = img2.astype(float)

    # Calculate squared difference
    squared_difference = (img1 - img2) ** 2

    # Calculate Mean Squared Error
    mse = np.mean(squared_difference)

    # Print MSE
    print("Mean Squared Error:", mse)

    # Display both images
    plt.figure(figsize=(7, 5))

    plt.subplot(1, 2, 1)
    plt.imshow(img1, cmap="gray")
    plt.title("Image 1")
    plt.axis("off")

    plt.subplot(1, 2, 2)
    plt.imshow(img2, cmap="gray")
    plt.title("Image 2")
    plt.axis("off")

    plt.tight_layout()
    plt.show()

else:
    print("Image not found. Check the file paths.")
```
